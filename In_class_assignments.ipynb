```python
#Assignment number two
num_1=int(input("enter the first number"))
num_2=int(input("enter the second number"))
op_1=input("enter the operator")
if op_1=="+":
    print(num_1+num_2)
elif op_1=="-":
  print(num_1-num_2)
elif op_1=="*":
  print(num_1*num_2)
elif op_1=="/":
  print(num_1/num_2)
elif op_1=="//":
  print(num_1//num_2)
elif op_1=="**":
  print(num_1**num_2)
else:
  print("invalid operator")
```

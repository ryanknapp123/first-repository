```python
#loan & npv mini-toolkit(exceptions)
#partA
def roi(cost, revenue):
    assert isinstance(revenue, (int, float)), "revenue must be numeric"
    assert isinstance(cost, (int, float)), "cost must be numeric"
    assert cost != 0, "cost cannot be zero"
    roi_value = float(revenue - cost) / cost
    assert isinstance(roi_value, float), "ROI must be a float"
    return roi_value

try:
    cost_input = float(input("enter the cost of the loan: "))
    rev_input = float(input("enter the revenue of the loan: "))
    print(roi(cost_input, rev_input) * 100, "%")
except (ValueError, AssertionError) as e:
    print(f"Error: {e}")

def loan_payment(principal, annual_rate, months):
    assert isinstance(principal, (int, float)), "Principal must be numeric"
    assert isinstance(annual_rate, (int, float)), "Annual rate must be numeric"
    assert isinstance(months, (int, float)), "Months must be numeric"

    if principal <= 0 or annual_rate <= 0 or months <= 0:
        raise ValueError("All inputs must be greater than 0")

    r = (annual_rate / 100) / 12
    if r != 0:
        pmt = (r * principal) / (1 - (1 + r)**(-months))
        return pmt
    else:
        return principal / months
```

```python
class Employee:
  def __init__(self, name, hourly_wage, hours_worked):
    self.name = name
    self.__hourly_wage = hourly_wage
    self.hours_worked = hours_worked

  def calculate_pay(self):
    return self.__hourly_wage * self.hours_worked

  def get_wage(self):
    return self.__hourly_wage

  def set_wage(self, new_wage):
    if new_wage < 0:
      raise ValueError("Hourly wage cannot be negative")
    self.__hourly_wage = new_wage

  def __str__(self):
    return f"Employee Name: {self.name}, Hourly Wage: ${self.__hourly_wage:.2f}, Hours Worked: {self.hours_worked}"

class Manager(Employee):
  def __init__(self, name, hourly_wage, hours_worked, bonus=0):
    super().__init__(name, hourly_wage, hours_worked)
    self.bonus = bonus

  def calculate_pay(self):
    return super().calculate_pay() + self.bonus

  def __str__(self):
    return f"Manager Name: {self.name}, Hourly Wage: ${self.__hourly_wage:.2f}, Hours Worked: {self.hours_worked}, Bonus: ${self.bonus:.2f}"

employee1 = Employee("Alice Smith", 15.00, 40)

manager1 = Manager("Bob Johnson", 25.00, 40, 500)

print("--- Employee Information ---")
print(employee1)
print(f"Weekly Pay: ${employee1.calculate_pay():.2f}")

print("\n--- Manager Information ---")
print(manager1)
print(f"Weekly Pay: ${manager1.calculate_pay():.2f}")

print("\n--- Employee Wage Update ---")
employee1.set_wage(17.50)
print(f"New Hourly Wage for {employee1.name}: ${employee1.get_wage():.2f}")
print(f"New Weekly Pay: ${employee1.calculate_pay():.2f}")
```

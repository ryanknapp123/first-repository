```python
#Assignment number two
num_1=int(input("enter the first number"))
num_2=int(input("enter the second number"))
op_1=input("enter the operator")
if op_1=="+":
    print(num_1+num_2)
elif op_1=="-":
  print(num_1-num_2)
elif op_1=="*":
  print(num_1*num_2)
elif op_1=="/":
  print(num_1/num_2)
elif op_1=="//":
  print(num_1//num_2)
elif op_1=="**":
  print(num_1**num_2)
elif op_1=="%":
  print(num_1%num_2)
else:
  print("invalid operator")
```

```python
#assignment 3
fee=2.5
flat_rate=5
total_charge=0
n=0
while n<=8:
  hours=int(input("enter the number of hours you were parked for. the max park time is 8 hours"))
  n=hours
  if n>0 and n<=8:
    cost=fee*n+flat_rate
    print("your fee is",cost,"hours parked=",n)
    total_charge+=cost
    if total_charge > 20:
      total_charge = 20
      print("Maximum charge of $20 reached. Final total charge:", total_charge)
      break
    elif total_charge <= 10:
      total_charge = 10
      print("Total charge set to minimum of $10. Final total charge:", total_charge)
      break
  elif n==0:
    print("No hours entered. Exiting.")
    break
  else:
    print("Invalid hours. Max park time is 8 hours. Please try again.")
```
